## 82 — Další metody hráče


```python
class HracSPravidly:
    maximum_zivotu = 3

    def __init__(self, jmeno):
        self.jmeno = jmeno
        self.zivoty = self.maximum_zivotu

    def popis(self):
        return f"{self.jmeno}: {self.zivoty}/{self.maximum_zivotu}"

    @staticmethod
    def platna_rychlost(rychlost):
        return rychlost >= 0

    @classmethod
    def vychozi(cls):
        return cls("Nový hráč")

alex = HracSPravidly("Alex")
alex.zivoty -= 1
print(alex.popis())
print("Platná rychlost:", HracSPravidly.platna_rychlost(5))
novy_hrac = HracSPravidly.vychozi()
print("Ve dvou krocích:", novy_hrac.popis())
print("Třídní konstrukční metoda:", HracSPravidly.vychozi().popis())
alex.maximum_zivotu = 5
print("Překrytí v instanci:", alex.maximum_zivotu)
print("Atribut třídy:", HracSPravidly.maximum_zivotu)
```

## 84 — Další dědičnost: kouzelníci


```python
class HerniPostava:
    def __init__(self, jmeno):
        self.jmeno = jmeno
        self.zivoty = 3

    def popis(self):
        return f"{self.jmeno}, životy {self.zivoty}"

class Kouzelnik(HerniPostava):
    def __init__(self, jmeno, mana):
        super().__init__(jmeno)
        self.mana = mana

    def popis(self):
        return f"{super().popis()}, mana {self.mana}"

class MistrKouzelnik(Kouzelnik):
    def kouzli(self):
        return f"{self.jmeno} sesílá kouzlo"

print(HerniPostava("Alex").popis())
print(Kouzelnik("Bára", 10).popis())
mistr = MistrKouzelnik("Cyril", 20)
print(mistr.popis())
print(mistr.kouzli())
```

## 86 — Vektor a speciální metody


```python
class Vektor:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, dalsi):
        if not isinstance(dalsi, Vektor):
            return NotImplemented
        return Vektor(self.x + dalsi.x, self.y + dalsi.y)

    def __str__(self):
        return f"Vektor({self.x}, {self.y})"

pozice = Vektor(10, 20)
posun_vektor = Vektor(5, -2)
print("Nový vektor:", pozice + posun_vektor)
print("Původní vektor:", pozice)
```

## 94–98 — Řešení 7: zpracování pohybů


```python
def zpracuj_pohyby(x, prikazy, rychlost=5):
    for prikaz in prikazy:
        if prikaz == "konec":
            break
        if prikaz == "vpravo":
            x += rychlost
        elif prikaz == "vlevo":
            x -= rychlost
    return x

prikazy = ["vpravo", "vpravo", "vlevo", "konec", "vpravo"]
print("Konečné x:", zpracuj_pohyby(10, prikazy))
assert zpracuj_pohyby(10, prikazy) == 15
assert zpracuj_pohyby(10, ["neznamy"]) == 10
assert zpracuj_pohyby(10, []) == 10
assert zpracuj_pohyby(10, ["konec", "vpravo"]) == 10
assert zpracuj_pohyby(10, ["vlevo"], rychlost=0) == 10

assert zpracuj_pohyby(10, ["vpravo", "vpravo"]) == 20
assert zpracuj_pohyby(10, ["vlevo", "vlevo"]) == 0
assert zpracuj_pohyby(10, ["neznamy", "vpravo"]) == 15

print(zpracuj_pohyby(10, ["vpravo", "vpravo", "vlevo"]))
print(zpracuj_pohyby(10, ["konec", "vpravo"]))
print(zpracuj_pohyby(10, ["cekej"]))
print(zpracuj_pohyby(10, ["vpravo", "cekej"]))
```
